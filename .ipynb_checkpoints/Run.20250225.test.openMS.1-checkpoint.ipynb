```python
from urllib.request import urlretrieve
import pyopenms as oms

exp = oms.MSExperiment()
# Load mzML file and obtain spectrum for peptide YIC(Carbamidomethyl)DNQDTISSK
oms.MzMLFile().load(mzML_path, exp)

# Get first spectrum
spectra = exp.getSpectra()
observed_spectrum = spectra[0]
```

```python
tsg = oms.TheoreticalSpectrumGenerator()
theo_spectrum = oms.MSSpectrum()
p = tsg.getParameters()
p.setValue("add_y_ions", "true")
p.setValue("add_b_ions", "true")
p.setValue("add_metainfo", "true")
tsg.setParameters(p)
peptide = oms.AASequence.fromString("YIC(Carbamidomethyl)DNQDTISSK")
tsg.getSpectrum(theo_spectrum, peptide, 1, 2)
print(theo_spectrum)
```

```python
import matplotlib.pyplot as plt
from pyopenms.plotting import mirror_plot_spectrum
import matplotlib.pyplot as plt

mirror_plot_spectrum(
    observed_spectrum,
    theo_spectrum,
    spectrum_bottom_kws={"annotate_ions": False},
)
plt.show()
```

```python
alignment = []
spa = oms.SpectrumAlignment()
p = spa.getParameters()
# use 0.5 Da tolerance (Note: for high-resolution data we could also use ppm by setting the is_relative_tolerance value to true)
p.setValue("tolerance", 0.5)
p.setValue("is_relative_tolerance", "false")
spa.setParameters(p)
# align both spectra
spa.getSpectrumAlignment(alignment, theo_spectrum, observed_spectrum)
```

```python
from tabulate import tabulate


# Print matching ions and mz from theoretical spectrum
print("Number of matched peaks: " + str(len(alignment)))
t = []
for theo_idx, obs_idx in alignment:
    ion_name = theo_spectrum.getStringDataArrays()[0][theo_idx].decode()
    ion_charge = theo_spectrum.getIntegerDataArrays()[0][theo_idx]
    t.append(
        [
            ion_name,
            str(ion_charge),
            str(theo_spectrum[theo_idx].getMZ()),
            str(observed_spectrum[obs_idx].getMZ()),
        ]
    )
print(tabulate(t, headers=["ion", "charge", "theo. m/z", "observed m/z"]))
```

```python
import pyopenms as oms

# 定义文件路径
mzML_path = "/rd1/user/lit/project/sORFs/raw_data/MS/Guomics_SP_MSdata/CAD20241224licq_BSEP_DDA_60min/CAD20241224licq_BSEP_DDA_60min_E16_3_3-30K_LC-T_Slot2-18_1_4691_calibrated.mzML"
mgf_path = "/rd1/user/lit/project/sORFs/raw_data/MS/Guomics_SP_MSdata/CAD20241224licq_BSEP_DDA_60min/CAD20241224licq_BSEP_DDA_60min_E16_3_3-30K_LC-T_Slot2-18_1_4691_uncalibrated.mgf"

```

```python
# 加载 mzML 文件（实验数据）
exp = oms.MSExperiment()
oms.MzMLFile().load(mzML_path, exp)
experimental_spectra = exp.getSpectra()

# 加载 mgf 文件（理论数据）
theo_exp = oms.MSExperiment()
oms.MascotGenericFile().load(mgf_path, theo_exp)
theoretical_spectra = theo_exp.getSpectra()

```

```python
# 假设我们比较第一个实验光谱和第一个理论光谱
observed_spectrum = experimental_spectra[0]
theo_spectrum = theoretical_spectra[0]

# 对齐光谱
alignment = []
spa = oms.SpectrumAlignment()
p = spa.getParameters()
p.setValue("tolerance", 0.5)  # 设置质量误差容限（0.5 Da）
p.setValue("is_relative_tolerance", "false")  # 使用绝对误差
spa.setParameters(p)
spa.getSpectrumAlignment(alignment, theo_spectrum, observed_spectrum)

# 提取匹配峰和非匹配峰
matched_pairs = []
unmatched_pairs = []

# 获取匹配的峰
for theo_idx, obs_idx in alignment:
    mz_theo = theo_spectrum[theo_idx].getMZ()
    intensity_theo = theo_spectrum[theo_idx].getIntensity()
    mz_obs = observed_spectrum[obs_idx].getMZ()
    intensity_obs = observed_spectrum[obs_idx].getIntensity()
    matched_pairs.append((mz_theo, intensity_theo, mz_obs, intensity_obs))

# 获取非匹配的峰
# 遍历实验光谱中的所有峰，检查是否在匹配结果中
for i in range(len(observed_spectrum)):
    mz_obs = observed_spectrum[i].getMZ()
    intensity_obs = observed_spectrum[i].getIntensity()
    
    # 检查是否在匹配结果中
    is_matched = False
    for _, obs_idx in alignment:
        if i == obs_idx:
            is_matched = True
            break
    
    if not is_matched:
        unmatched_pairs.append((mz_obs, intensity_obs))

# 打印匹配峰和非匹配峰
print("Matched Peaks (m/z_theo, intensity_theo, m/z_obs, intensity_obs):")
for pair in matched_pairs:
    print(pair)

print("\nUnmatched Peaks (m/z_obs, intensity_obs):")
for pair in unmatched_pairs:
    print(pair)

# 保存结果到文件
with open("matched_peaks.txt", "w") as f:
    f.write("m/z_theo\tintensity_theo\tm/z_obs\tintensity_obs\n")
    for pair in matched_pairs:
        f.write(f"{pair[0]}\t{pair[1]}\t{pair[2]}\t{pair[3]}\n")

with open("unmatched_peaks.txt", "w") as f:
    f.write("m/z_obs\tintensity_obs\n")
    for pair in unmatched_pairs:
        f.write(f"{pair[0]}\t{pair[1]}\n")
```
