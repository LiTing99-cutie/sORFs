```python
import os
from pyteomics import mgf, pepxml, mass, pylab_aux
from urllib.request import urlopen, Request
import pylab

def get_single_spectrum(title,title_1, psms, mgf_path,output_file):
    psm=psms[title]
    
    # Retrieve the spectrum using the mgf path and title
    spectrum = mgf.get_spectrum(mgf_path, title_1)

    # Create a new figure for the plot
    plt.figure(figsize=(10, 6))

    # Annotate the spectrum
    pylab_aux.annotate_spectrum(
        spectrum,
        psm['search_hit'][0]['peptide'],
        title='Annotated spectrum ' + psm['search_hit'][0]['peptide'],
        maxcharge=psm['assumed_charge']
    )

    # Save the annotated spectrum as a PDF
    plt.savefig(output_file, format='pdf')
    plt.close()  # Close the figure to free up memory

# Example usage:
# get_single_spectrum(title, pep_xml_path, mgf_path, psm)
```

# 整理为函数

```python
import os
from pyteomics import mgf, pepxml, mass, pylab_aux
from urllib.request import urlopen, Request
import pylab
import pandas as pd
import glob
import matplotlib.pyplot as plt
```

```python
def get_single_spectrum(title,title_1, psms, mgf_path,output_file,Format="pdf"):
    psm=psms[title]
    
    # Retrieve the spectrum using the mgf path and title
    spectrum = mgf.get_spectrum(mgf_path, title_1)

    # Create a new figure for the plot
    plt.figure(figsize=(10, 6))

    # Annotate the spectrum
    pylab_aux.annotate_spectrum(
        spectrum,
        psm['search_hit'][0]['peptide'],
        title='Annotated spectrum ' + psm['search_hit'][0]['peptide'],
        maxcharge=psm['assumed_charge']
    )

    # Save the annotated spectrum as a PDF
    plt.savefig(output_file, format=Format)
    plt.close()  # Close the figure to free up memory
```

```python
def get_batch_spectrum(file_path,output_path,Fragpipe_output_path, rawdata_path,Format):
    df = pd.read_csv(file_path, sep='\t', header=0)
    df.columns = ['SpectrumTitle', "SpectrumFile",'Sample_Name',"SpectrumTitle_1"]
    # 遍历数据框，处理每个条目
    for index, row in df.iterrows():
        title = row['SpectrumTitle']
        title_1 = row['SpectrumTitle_1']
        Sample_Name = row['Sample_Name']

        sample_prefix=title.split('.')[0]
        mgf_path=rawdata_path+sample_prefix+"_uncalibrated.mgf"
        pep_xml_path_1=Fragpipe_output_path+Method+"/Trypsin/"+sample_prefix.replace("-", "_")+"_d/"+"*pep.xml"
        pep_xml_path=glob.glob(pep_xml_path_1)[0]
        psms=pepxml.read(pep_xml_path)

        # 创建以蛋白质名字命名的文件夹路径
        protein_folder = output_path+"/"

        # 如果文件夹不存在，则创建它
        if not os.path.exists(protein_folder):
            os.makedirs(protein_folder)

        # 构建完整的文件路径
        file_path = os.path.join(protein_folder, title + "."+Format)

        # 调用函数处理数据并生成输出文件
        get_single_spectrum(title, title_1,psms, mgf_path, file_path,Format)
    print("处理完成。")
```
