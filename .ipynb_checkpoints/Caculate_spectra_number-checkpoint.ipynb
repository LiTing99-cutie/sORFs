```r
get_ms2_count <- function(MZML_FILE){
    library("Spectra")
    ms <- Spectra(MZML_FILE)
    Sample <- str_extract(MZML_FILE, "(?<=min_)\\w+(?=_Slot)")
    ms2_count <- sum(msLevel(ms) == 2)
    df <- data.frame(Sample=Sample,ms2_count=ms2_count)
    return(df)
}
```

```r
lapply(mzml,get_ms2_count) %>% do.call(rbind,.)
```
