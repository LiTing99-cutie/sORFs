```r
unfold_nonUnique <- function(df){
    # 分组并且展开
    df$Group_id <- 1:nrow(df)
    df %>% separate_rows(`Mapped.Proteins`, convert = TRUE,sep=", ") -> tmp
    tmp %>% mutate(Protein=`Mapped.Proteins`,
                   `Protein.ID`=`Mapped.Proteins`,
                   Gene=`Mapped.Proteins`,
                   `Entry.Name`=`Mapped.Proteins`,
                   `Protein.Description`=`Mapped.Proteins`) -> tmp_1
    rbind(df,tmp_1) -> df_unfold
    return(df_unfold)
}
```

```r
psm %>% filter(grepl("ENS",Protein)) %>% filter(Is.Unique=="FALSE" | Is.Unique=="false") %>% 
  filter(!grepl("sp",Mapped.Proteins)) -> psm_non_unique
```

```r
psm_non_unique -> df
unfold_nonUnique(df) -> df_unfold
nrow(df_unfold)
```

```r
# 存在一些虽然Is.Unique为FALSE，但是Mapped.Proteins为空的记录
df_unfold[!grepl("^ENS",df_unfold$Protein),] %>% head()
df$Group_id <- 1:nrow(df)
filter(df,Group_id==155)
# 把这些展开后为空的记录移除
df_unfold %>% filter(Protein!="") -> df_unfold
nrow(df_unfold)
```
