```r
# 修改了样本名匹配规则
get_ms2_count <- function(MZML_FILE){
    library("Spectra")
    ms <- Spectra(MZML_FILE)
    Sample <- str_extract(MZML_FILE, "(?<=min_).*?(?=_Slot)")
    ms2_count <- sum(msLevel(ms) == 2)
    df <- data.frame(Sample=Sample,ms2_count=ms2_count)
    return(df)
}
```

```r
get_ms2_count("/rd1/user/lit/project/sORFs/raw_data/MS/Guomics_SP_MSdata/CAD20241120licq_BSEP_C8_PCP_DDA_60min/CAD20241120licq_BSEP_DDA_60min_m14_C8_Slot2-1_1_4073_uncalibrated.mzML")
```

```r
lapply(mzml,get_ms2_count) %>% do.call(rbind,.) -> sample_ms2_count
```

```r
sample_ms2_count
```

```r
sample_ms2_count$Sample <- gsub(pattern = "-",replacement = "_",x = sample_ms2_count$Sample)
```
